# Question 1

```python
import pandas as pd

# Load the dataset
ts_songs = pd.read_csv('/anvil/projects/tdm/data/spotify/taylor_swift_discography_updated.csv', sep=';')
pd.set_option('display.max_columns', None)

ts_songs.head()
```

```python
# 1.1
# maximum energy level
max_energy = ts_songs['energy'].max()
print(f"The maximum energy level is: {max_energy}")

# minimum energy level
min_energy = ts_songs['energy'].min()
print(f"The minimum energy level is: {min_energy}")
```

```python
def find_songs_with_energy(input_df, threshold):
    my_output = input_df[input_df["energy"] >= threshold]
    return my_output

# median energy
my_median = ts_songs['energy'].median()
# 1.2 songs >= median 
high_energy_df = find_songs_with_energy(ts_songs, my_median)
print(f"There are {len(high_energy_df)} songs with energy greater than or equal to the median.")
```

```python
# 1.3 
def find_songs_by_album(input_df, album_name):
    album_songs = input_df[input_df["album"] == album_name]
    return album_songs

# test func
ttpd_anthology = find_songs_by_album(ts_songs, "The Tortured Poets Department: The Anthology")
print(f"Number of rows for the anthology: {len(ttpd_anthology)}")
ttpd_anthology.head()
```

# Question 2 

```python
# 2.1 
# milliseconds to seconds
ts_songs['duration_sec'] = ts_songs['duration_ms'] / 1000
# seconds to minutes
ts_songs['duration_min'] = ts_songs['duration_sec'] / 60
ts_songs[['track_name', 'duration_ms', 'duration_min']].head()
```

```python
def find_albums_by_time(input_df, input_min_duration):
    total_by_album = input_df.groupby("album")["duration_min"].sum().reset_index()
    long_albums = total_by_album[total_by_album["duration_min"] > input_min_duration]
    return long_albums

# Test 2.2 -> albums longer than 80 minutes
albums_80 = find_albums_by_time(ts_songs, 80)
print("Albums longer than 80 minutes:")
print(albums_80)

# Test 2.3 -> albums longer than 120 minutes
albums_120 = find_albums_by_time(ts_songs, 120)
print("\nAlbums longer than 120 minutes:")
print(albums_120)
```

# Question 3

```python
# 3.1 set the lower/upper album length thresholds
lower_threshold = 30
upper_threshold = 80
```

```python
# 3.2
def albums_by_length(df, lower_threshold, upper_threshold):
    total_by_album = df.groupby("album")["duration_min"].sum().reset_index()
    def classify(duration):
        if duration < lower_threshold:
            return "Short"
        elif duration > upper_threshold:
            return "Long"
        else:
            return "Medium"
    total_by_album["category"] = total_by_album["duration_min"].apply(classify)
    return total_by_album

# 3.3
final_albums_df = albums_by_length(ts_songs, lower_threshold=30, upper_threshold=80)
print(final_albums_df)
```

# Question 4

```python
youtubers = pd.read_csv('/anvil/projects/tdm/data/youtube/most_subscribed_youtube_channels.csv')

# 4.1 make numeric and clean
youtubers['video_count2'] = youtubers['video count'].str.replace(",", "", regex=False)
youtubers['video_count2'] = pd.to_numeric(youtubers['video_count2'])
youtubers[['Youtuber', 'video count', 'video_count2']].head()
```

```python
# 4.2 count per category
category_video_counts = youtubers.groupby('category')['video_count2'].sum()

print("Total videos per category:")
print(category_video_counts)
```

```python
# 4.3 show grouped_counts
gm_subset = youtubers[youtubers["category"].isin(["Gaming", "Music"])]
# group by category + started, sum and unstack
grouped_counts = gm_subset.groupby(["category", "started"])["video_count2"].sum().unstack()
print("Grouped counts (Gaming vs Music by Year):")
grouped_counts
```

# Question 5

```python
# 5.1 
# clean the subscribers column
youtubers['subscribers2'] = youtubers['subscribers'].str.replace(",", "", regex=False)
youtubers['subscribers2'] = pd.to_numeric(youtubers['subscribers2'])

def top_youtuber_by_genre(df, genre):
    genre_rows = df[df["category"] == genre]
    top_index = genre_rows['subscribers2'].idxmax()
    return genre_rows.loc[top_index]

# 5.1
top_gaming = top_youtuber_by_genre(youtubers, "Gaming")
print(top_gaming)
```

```python
# 5.2
top_music = top_youtuber_by_genre(youtubers, "Music")
print(top_music[['Youtuber', 'subscribers2']])
```
