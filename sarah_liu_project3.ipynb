```python
import pandas as pd
```

# Q1

```python
video_games= pd.read_csv('/anvil/projects/tdm/data/grouping/vgchartz-2024.csv')
video_games.head()
```

```python
# 1.1 group games by genre and calculate the mean of the critic score
genre = video_games.groupby("genre", as_index=False)["critic_score"].mean()
# sort then display top 5 genres with highest avg critic scores
genre.sort_values("critic_score", ascending=False).head()

# highest critic score is sandbox
```

```python
# 1.2 group by console and total num of sales
console = video_games.groupby("console", as_index=False)["total_sales"].sum()
# sort total sales then display consols with highest number of sales 
console.sort_values("total_sales", ascending=False).head()

# highest number of console sales is PS2
```

1.3 
Grouping data is a good way to organize large datasets into smaller, more specific categories. For example, in this case by genre or console. Then we can apply other functions like sum() or mean() to these groups to see trends in these specified categories instead of looking at the entire list at once. It's useful for summarizing thousands of rows of raw data into smaller, more manageable stuff.

# Q2

```python
# 2.1 high_rated column T if score >= 7
video_games['high_rated'] = video_games['critic_score'] >= 7
# low_rated column T = if score < 7
video_games['low_rated'] = video_games['critic_score'] < 7

# display the value counts for both
display(video_games['high_rated'].value_counts())
display(video_games['low_rated'].value_counts())

video_games['critic_score'].isna().sum()
```

```python
#2.1 filter df to genre = sports and highrated = true
sports_hits = video_games[(video_games['genre'] == 'Sports') & (video_games['high_rated'] == True)]
# display games with a critic score of 9.5
sports_hits[sports_hits['critic_score'] == 9.5]

# Games with critic score of exactly 9.5: Madden NFL 2004 and 2005, SSX 3
```

```python
# 2.3 sports failed dataframe 
sports_fails = video_games[(video_games['genre'] == 'Sports') & (video_games['low_rated'] == True)]
# display games w critic score of 2.8
sports_fails[sports_fails['critic_score'] == 2.8]

# games with critic score of exactly 2.8: Major League Baseball 2k10, MLB Slam!
```

```python
# 3.1 genre_counts = genre and num of titles in genre
genre_counts = video_games.groupby("genre", as_index=False)["title"].size()
# rename coulumn to game_count
genre_counts = genre_counts.rename(columns={"size": "game_count"})
# sort by number of games 
genre_counts = genre_counts.sort_values(by="game_count", ascending=False) 
```

```python
# 3.2 show first 10 rows 
genre_counts.head(10) 

# Misc genre has most games listed in dataset 
```

```python
# 3.3 grouping console + genre -> use .size() - get num of games for each combo
console_genre_counts = video_games.groupby(["console", "genre"], as_index=False).size()
# show first 5 rows 
console_genre_counts.head()
```

```python
app_store = pd.read_csv('/anvil/projects/tdm/data/grouping/googleplaystore.csv')
pd.set_option('display.max_columns', None)
```

```python
app_store["Clean_Reviews"] = app_store["Reviews"].str.replace(",", "", regex=False)
```

```python
# 3.1 'M' is causing the error 
```

```python
app_store["Numeric_Reviews"] = app_store["Clean_Reviews"].astype(float)
```

```python
# 4.2 remove 3.0 value
app_store = app_store[app_store["Clean_Reviews"] != "3.0M"]

# convert the column to float
app_store["Numeric_Reviews"] = app_store["Clean_Reviews"].astype(float)
```

```python
# new column for rating per review = rating / num reviews
app_store['rating_per_review'] = app_store['Rating'] / app_store['Numeric_Reviews']

# 4.3 display head
app_store['rating_per_review'].head()
```
