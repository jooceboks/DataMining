```python
# read dataset - pandas 
start = time.time()
pandas_df = pd.read_csv('/anvil/projects/tdm/data/flights/subset/1987.csv')
end = time.time()
print(f"Pandas read_csv took {end - start} seconds")

# read dataset - polars
start = time.time()
polars_df = pl.read_csv('/anvil/projects/tdm/data/flights/subset/1987.csv', null_values=["NA"])
end = time.time()
print(f"Polars read_csv took {end - start} seconds")
```

```python
# for pandas
start = time.time()
pandas_df = pd.read_csv('/anvil/projects/tdm/data/lahman/data/AllstarFull.csv', header=None)
pandas_df.columns = ['playerID', 'yearID', 'gameNum', 'gameID', 'teamID', 'lgID', 'GP', 'startingPos']
end = time.time()
print(f"Pandas read_csv took {end - start} seconds")

# for polars
start = time.time()
polars_df = pl.read_csv('/anvil/projects/tdm/data/lahman/data/AllstarFull.csv', has_header=False)
polars_df.columns = ['playerID', 'yearID', 'gameNum', 'gameID', 'teamID', 'lgID', 'GP', 'startingPos']
end = time.time()
print(f"Polars read_csv took {end - start} seconds")
```

```python
# 2.1 select columns playID and teamID in pandas/polars
# added yearID for part2.3
players_pandas = pandas_df[['playerID', 'teamID', 'yearID']]
players_polars = polars_df.select(pl.col("playerID"), pl.col("teamID"), pl.col("yearID"))
print(players_pandas.head())
print(players_polars.head())
```

```python
# find the PHI teamID entries in pandas_df
philly_players_pandas = players_pandas[pandas_df['teamID'] == "PHI"]
# filter polars_df by the teamID column to find the PHI entries
philly_players_polars = players_polars.filter(pl.col("teamID") == "PHI")

# 2.2 print head of philly players
print(philly_players_pandas.head())
print(philly_players_polars.head())
```

```python
# 2.3 sort by yearID = 2023
pandas2023 = philly_players_pandas[philly_players_pandas['yearID'] == 2023]
polars2023 = philly_players_polars.filter(pl.col("yearID") == 2023)

print(pandas2023.head())
print(polars2023.head())
```

# Q3
3.1 polars uses methods while pandas jsut uses brackets [] notation. Polars also uses regular expression making it more efficient for large datasets with more columns.

```python
# read 2005 flights w polars
flights05 = pl.read_csv("/anvil/projects/tdm/data/flights/subset/2005.csv")

# 3.2 create df for flights listed as "IND"
indy_flights = flights05.filter(pl.col("Origin") == "IND")

# 3.3 select columns starting with "CRS" + display
crs = indy_flights.select(pl.col("^CRS.*$"))
print(crs)
```

# Q4

```python
#4.1 there are 7 columns with the word Delay in them 
# find cols with word Delay in them - starting, in middle or at end of string
delay_cols = indy_flights.select(pl.col("^.*Delay.*$"))
print(delay_cols)
```

```python
arr_dep= indy_flights.select(pl.col("^(Arr|Dep).*$"))
print(arr_dep.head())
```

```python
# 4.3 ArrDelay and DepDelay cols 
delay_cols = indy_flights.select(pl.col("^(Arr|Dep).*Delay$"))
print(delay_cols)
```
