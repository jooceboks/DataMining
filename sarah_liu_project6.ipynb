# Question 1

```python
import pandas as pd
# read dataset
myDF = pd.read_csv('/anvil/projects/tdm/data/flights/subset/2006.csv')
myDF.head() # display head
```

```python
# 1.1 
pa_flights = myDF[myDF['Dest'] == 'PHL'] # filter for flights to PHL
PHL_origins = pa_flights['Origin'].nunique() # find num of unique origin spots
print(f"Origins flights in PHL 2006: {PHL_origins}") # display num of origins glights in PHL 2006
```

```python
# 1.2 
# group by origin and gind avg airtime 
PHL_avgAT = pa_flights.groupby('Origin')['AirTime'].mean().sort_values(ascending=False)
SFO_AT = PHL_avgAT['SFO'] # find avg airtime SFO + display results 
print(f"average AirTime for flights leaving SFO in pa_flights in 2006: {SFO_AT:.3f}")
```

```python
#1.3 find avg airtime HNL in az flights + print result
az_flights = myDF[myDF['Dest'] == 'PHX']
PHX_avgAT = az_flights.groupby('Origin')['AirTime'].mean().sort_values(ascending=False)
HNL_AT = PHX_avgAT['HNL'] 
print(f"average AirTime for flights leaving HNL in az_flights in 2006: {HNL_AT:.3f}")
```

```python
ice_cream = pd.read_csv('/anvil/projects/tdm/data/icecream/combined/products.csv')
# 3.1 
print(ice_cream['rating'].describe())
```

```python
# 3.2 display ranges of qcut before applying labels 
print(pd.qcut(ice_cream['rating'], q=4).value_counts().sort_index())
```

```python
# applying labels 
labels = ["Wouldn't Recommend", "Needs Improvement", "Solid Choice", "Fan Favorite"]
# create new column 'rating_phrase' using qcut w/ labels
ice_cream['rating_phrase'] = pd.qcut(ice_cream['rating'], q=4, labels=labels)

# 3.3 show first 5 rows
print(ice_cream.head())
```

# Question 4

```python
my_flights = pd.read_csv('/anvil/projects/tdm/data/flights/subset/1997.csv')
```

```python
# 4.1 define time ranges
time_bins = [0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400]

# define labels for each time bin
time_labels = ["Late Night", "Early Morning", "Morning", "Late Morning", 
               "Afternoon", "Early Evening", "Evening", "Night"]

# 4.1 create column depart_times column + print
my_flights['depart_times'] = pd.cut(my_flights['DepTime'], bins=time_bins, labels=time_labels)

# 4.2 create columnarrival_times column + print
my_flights['arrival_times'] = pd.cut(my_flights['ArrTime'], bins=time_bins, labels=time_labels)

print(my_flights)
```

```python
# show counts for flights (dep time and arrival time) for all time periods 
print(my_flights['depart_times'].value_counts())
print(my_flights['arrival_times'].value_counts())
# Departure time was busiest in the mornings and arrival times was busiest in the early evenings
```

# Question 5 

```python
# select rows for Feb (Month == 2)
Feb = my_flights[my_flights["Month"] == 2]
# value counts for DayOfWeek in Feb
print(Feb["DayOfWeek"].value_counts().sort_index())
# value counts for DayOfWeek in all months 
print(my_flights["DayOfWeek"].value_counts().sort_index())
```

```python
# 5.2 dataframe total airtime across days and months
total_airtime = my_flights.groupby(["DayOfWeek", "Month"])["AirTime"].sum().unstack()
print(total_airtime)
```

```python
# 5.3 avg AirTime across days and months
avg_airtime = my_flights.groupby(["DayOfWeek", "Month"])["AirTime"].mean().unstack()
print(avg_airtime)
```
