```python
# 3.1 screen time loop
screen_time = 10
while screen_time > 2:
    print(f"Screen time: {screen_time} hours")
    screen_time = screen_time - 1
print(f"Final screen time: {screen_time}")
```

```python
# 3.1 sleep time loop
sleep_time = 2
while sleep_time < 10:
    print(f"Sleep time: {sleep_time} hours")
    sleep_time = sleep_time + 1
print(f"Final sleep time: {sleep_time}")
```

```python
# 3.2 loop increment/decrement = 0.5 + display
screen_time = 10
sleep_time = 2

while screen_time > 2:
    print(f"Screen Time: {screen_time}, Sleep Time: {sleep_time}")
    screen_time -= 0.5
    sleep_time += 0.5

print(f"Final values Screen: {screen_time}, Sleep: {sleep_time}")
```

# Question 4

```python
# 4.1 Load the dataset + show specied of first 20 characters
# 4.2 add counters human or non human to count characters in first 20
characters = pd.read_csv('/anvil/projects/tdm/data/starwars/characters.csv')
i = 0
char_count = 0
human_count = 0
nonhuman_count = 0

while char_count < 20:
    if characters.loc[i, "species"] == "Human":
        print(f"Character number {i} ({characters.loc[i, 'name']}) is a human")
        human_count += 1
    else:
        print(f"Character number {i} ({characters.loc[i, 'name']}) is NOT a human")
        nonhuman_count += 1    
    i += 1
    char_count += 1
```

```python
# 4.3 Results
print(f"Total Humans: {human_count}")
print(f"Total NonHumans: {nonhuman_count}")
```
