```python
# Question 3
myDf['Status'] = myDf['Survived'].map({0: "Dead", 1: "Alive"})

# 3.1 show the value counts for status
print(myDf['Status'].value_counts())
```

```python
# 3.2 Combine status + sex w/ ,
myDf['Combined'] = myDf['Status'].str.cat(myDf['Sex'], sep=", ")

#display
print(myDf['Combined'].value_counts())
```

```python
# Question 4 
# 4.1 get youngest and oldest passenger and show
print('youngest:', myDf['Age'].min())
print('oldest:', myDf['Age'].max())
```

```python
# list of 'old ages'
old_ages = [61, 62, 63, 64, 65, 66, 70, 70.5, 71, 74, 80]

# .isin() to find 'old ages' in the Age column
old = myDf[myDf["Age"].isin(old_ages)]["Age"]

# 4.2 show count of ppl of old age
print(old.value_counts())
```

```python
# ages <= 10
my_list1 = list(myDf[myDf['Age'] <= 10]['Age'])
# ages > 10 and <= 60
my_list2 = list(myDf[(myDf['Age'] > 10) & (myDf['Age'] <= 60)]['Age'])

# 4.3 'people' = list1 + list2
people = my_list1 + my_list2
display(len(people)) # number of ppl <= 10 + > 10 and <=60
```

```python
# Question 5
# 5.1 list children, young, adult in age column
children = list(myDf[myDf['Age'] <= 20]['Age'])
young = list(myDf[(myDf['Age'] > 20) & (myDf['Age'] <= 40)]['Age'])
adult = list(myDf[(myDf['Age'] > 40) & (myDf['Age'] <= 60)]['Age'])

print(len(children)) # check list 
```

```python
# 5.2 creating age_group column
myDf.loc[myDf["Age"].isin(children), "AgeGroup"] = "Child"
myDf.loc[myDf["Age"].isin(young), "AgeGroup"] = "Young"
myDf.loc[myDf["Age"].isin(adult), "AgeGroup"] = "Adult"
myDf.loc[myDf["Age"].isin(old), "AgeGroup"] = "Old"

print(myDf["AgeGroup"].value_counts()) # check values in AgeGroup 
```

```python
# 5.3 using crosstab to cobine AgeGroup + Combined
age_analysis = pd.crosstab(myDf["AgeGroup"], myDf["Combined"])
display(age_analysis) # display 
```
