# Q4

```python
# 4.1 
df.loc[df['rating'] == 'R)', 'rating'] = 'R' # fix typos - 'R)' = 'R' rating 
df.loc[df['rating'] == 'PG-13)', 'rating'] = 'PG-13'  # 'PG-13)' = 'PG-13' rating
df['rating'].value_counts() #display 6 categories
```

```python
# 4.2 display tomatometer status w most films 
df['tomatometer_status'].value_counts().head(1)
# Rotten had the most films
```

```python
# 4.3 create cross table for rating and tomatometer status 
Td = pd.crosstab(df['rating'], df['tomatometer_status'])
display(Td) #display table
```
